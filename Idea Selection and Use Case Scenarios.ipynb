# Final Project Idea Selection and Use Case Scenarios (Milestone #2)

## Idea Pitch

KidTalk Mirror is an AI-powered insight tool that helps parents of 3–10-year-olds strengthen their child’s motivation, curiosity, and confidence simply by understanding their own words. Many parents want to connect but find themselves stuck in daily power struggles or emotional shutdowns. KidTalk Mirror acts as a gentle communication coach: parents record five minutes of real conversation, and the system, trained on developmental psychology research, transcribes and analyzes it to reveal key patterns like autonomy support, curiosity prompts, or pressure triggers. It then translates these insights into easy, actionable “language swaps” parents can try right away. By turning complex psychology into everyday guidance, KidTalk Mirror empowers families to build trust, resilience, and open communication with one small language shift at a time.


## Description of Target Users

1. The Academically-Focused Parent 
- Background: Parent of an elementary-aged child (6–10) in a competitive environment. They are highly invested in their child's academic success and future opportunities.
- Needs: Their child is resisting homework, seems unmotivated, or shuts down when facing challenges. They need to find a way to motivate their child without causing more conflict, tears, or anxiety.
- Goals: To see their child become more resilient, confident, and intrinsically motivated. They want to transform homework time from a daily battle into a more positive and productive interaction. 

2. The Connection-Focused Parent 
- Background: Parent of a 4–6-year-old who reads parenting Instagram, wants to be “emotionally supportive,” but still ends up saying “You’re fine, stop crying.” Feels guilty after conflict.
- Needs: Clear, non-judgy language swaps for tough moments (tantrums, bedtime resistance).
- Goals: Build emotional safety and trust so the child will talk to them, not push them away. 

3. Time-Stretched Working Parent 
- Background: Dual-working or single parent, limited patience at the end of the day. Most real conversations with the child happen during stressful routines (dinner, bedtime, homework).
- Needs: Fast, practical coaching without books, classes, or therapy.
- Goals: Fewer nightly battles, a calmer home, and a kid who can self-regulate instead of melting down

## More detailed summary of what the application will do

### Core Product Requirements

1. **Audio Capture and Upload:**  
   The system must allow a user to easily record a 5-10 minute audio clip (homework, bedtime, etc.) directly within the app or upload an existing audio file.

2. **Automated Transcription:**  
   The system must automatically and accurately transcribe the uploaded parent-child audio interaction into text with speaker separation (parent vs. child).

3. **Psychology/Cognitive Science-Based Analysis:**  
   The system must analyze the transcript to automatically identify and tag key communication patterns based on developmental psychology (e.g., autonomy-supportive vs. controlling language, shame triggers, curiosity prompts).

4. **Non-Judgmental Feedback:**  
   The system must present the analysis to the parent using gentle, supportive, and non-judgmental language, focusing on the impact of words rather than the intent of the parent.

5. **Actionable Language Swaps:**  
   The product must provide 2–3 concrete phrasing swaps the parent can try next time, tailored to what they actually said (e.g. “Instead of ___, try ___”).

6. **Secure and Private:**  
   The system must ensure all user-uploaded audio files and transcripts are encrypted, securely stored, and kept completely private to build user trust.

7. **Progress Tracking:**  
   The system must provide a simple dashboard for parents to visualize their communication patterns over time (e.g., “more calming language this week,” “fewer shame triggers”), so parents can see progress and stay motivated.

### Key Features and User Flows

**Feature 1: Interaction Analysis Report**

**Description:**  
Core product loop. It allows parents to either record a live interaction or upload an existing audio file (like a voice memo). The system then transcribes and analyzes the interaction, providing a simple, non-judgmental report that identifies key communication patterns and gives specific, actionable "language swaps.”

**User Flow:**
1. The user opens the app and taps the "New Report" button on the home screen.  
2. The app presents two clear options: "Record Live Moment" or "Upload Audio File.”  
3. If "Record": The user selects a context (e.g., "Homework," "Bedtime"), records 3-5 minutes, and presses “Stop."  
4. If "Upload": The user selects an existing audio file from their phone and is then prompted to select the context.  
5. The user confirms the submission. The app shows an "Analyzing your moment..." processing screen.  
6. A push notification arrives: "Your new Insight Report is ready!”  

**The user opens the report and sees:**
- *Gentle Summary:* A brief, encouraging overview.  
- *Key Patterns:* Simple tags identifying patterns like pressure (“Hurry up, just do it”), shame/comparison (“Why can’t you do this like other kids?”), emotional attunement (“This part feels hard, huh?”), and autonomy support (“Want to try it your way first?”).  
- *Actionable Swaps:* 2-3 cards showing their original quote and a concrete "Try This Instead" suggestion.

**Feature 2: Growth Dashboard**

**Description:**  
A private, personal dashboard that visualizes the parents' communication trends over time. This shifts the focus from a single "bad" interaction to a long-term "growth" journey, providing motivation and tracking progress.

**User Flow:**
1. The user taps the "Growth" tab from the main navigation.  
2. The screen displays simple, easy-to-read charts for the last 30 days.  
3. The user can toggle between charts for:  
   - "Helpful Patterns" (e.g., Curiosity Prompts, Emotional Attunement)  
   - "Patterns to Reduce" (e.g., Shame Triggers, Controlling Language)  
4. The dashboard also highlights "Weekly Wins" (e.g., "You used 30% more autonomy-supportive phrases this week!") to reinforce positive changes.

**Feature 3: "Skill Builder" Learning Hub**

**Description:**  
A curated, in-app library of "2-minute-read" articles, quick tips, and audio examples that proactively teach the core developmental psychology concepts behind the analysis (e.g., "What is Autonomy Support?," "3 Ways to Rephrase a ‘No'").

**User Flow:**
1. The user's Interaction Analysis Report (Feature 1) contains a link: "We tagged this as a 'Shame Trigger.' Learn more about shame vs. guilt."  
2. Clicking the link takes them to a short, simple article in the "Skill Builder" hub.  
3. Alternatively, the user can tap the "Learn" tab anytime.  
4. They can browse topics like "Building Motivation," "Managing Frustration," or "Encouraging Curiosity" to find quick, practical advice and examples to try before their next interaction.

## Data that will be needed

This project would require a large dataset of natural conversations between parents and their children in daily life in either audio or text format. We need both pre-training data to identify patterns and construct systematic prompts to train the model based on Cognitive Sciences and Education research, and post-training dataset to pilot and fine-tune the prompting. The data could be obtained from research studies in the field of education and children development. 


## Role of LLM

The prompted and fine-tuned LLMs are designed to act as a tutor with expertise in children development and education and give tailored suggestions on how the parents should communicate with their children properly. 


## How the product will differ? (from existing non-LLM powered systems in the domain and from just using a general purpose LLM for the task)

Right now, the existing non-LLM powered system in the domain connects parent users with human counselors in real-time, which is 1. Much more costly, 2. With more time-constraints since the users need to schedule a time with the counselors, 3. With uninsured qualities, because human evaluators can be inconsistent. Our product brings a more economic option to even underprivileged families with in-time support and systematic coaching strategies. 

In comparison to using a general purpose LLM for the task, our product 1. Provides an interface that makes the uploading and receiving processes much easier and user-friendly for the parents. 2. Provides research-backed, scientific coaching and suggestions to the parents based on system prompting, which most parents do not have the knowledge of. 


## Usage Scenarios

### Scenario 1: Homework Meltdown Check-in for Academically focused parents and their kid

Short description: Recording a stressful homework interaction and receiving real-time, psychology-based feedback.

1. Narrative of scenario: The user opens KidTalk Mirror during a frustrating math homework session with their kid and starts a live recording. The app captures a 5-minute dialogue between the parent and their 8-year-old. Once uploaded, the fine-tuned LLM (trained on thousands of parent-child communication transcripts and developmental psychology data) transcribes and analyzes the exchange. It identifies patterns like controlling language (“Just do it already”) and autonomy support (“Want to try solving this your way first?”). In the background, the model builds a pattern map that links each parental phrase to a predicted psychological impact score (pressure, encouragement, empathy). Using this internal structure, the LLM generates a gentle summary and personalized language swaps, focusing on promoting motivation and reducing stress triggers. The parent later reads their Insight Report, which visualizes these patterns and adds to their Growth Dashboard history.

3. Step-by-step interaction
- The user opens the app → taps “New Report → Record Live Moment.”
- Selects “Homework” as context.
- Records a 5-minute interaction; uploads automatically.
- LLM transcribes, detects speaker turns, and tags each utterance (e.g., pressure, empathy).
- The model generates a pattern map and constructs:
  1. Gentle Summary (“Homework can be challenging—you stayed engaged and calm.”)
  2. Tagged examples of controlling vs. autonomy-supportive phrases.
  3. 2–3 language swap cards with explanations.
- The user reviews the Insight Report and sees data logged in their dashboard.

3. Data description
- Needed data:
  1. Audio input: 5–10 minute recording of natural parent-child dialogue.
  2. Metadata: Context (“Homework”), time, anonymized user ID.
  3. Training data: Large pretraining corpus of parent-child conversations from educational research.
  4. Model resources: Psychological frameworks labeling autonomy, curiosity, and self-regulation patterns.
     
- Data created:
  1. Transcript + speaker separation between parents and children
  2. Pattern map: each utterance tagged with tone/impact category.
  3. Generated swaps for key phrases.\
  4. Weekly trend record integrated into user progress data.

4. Evaluation
- Success criteria: Tag accuracy ≥75%; user rates ≥1 swap as “Helpful.”
- Testing methods: Compare model tags to expert-coded transcripts.
- Complications: There may be background noise, misclassification of sarcasm languages. A safety mechanism can be designed to withhold or delay giving feedback if the model isn’t confident about its analysis.


### Scenario 2: Bedtime Tantrum Language Swap

Short description: User uploads a bedtime meltdown to reveal patterns of emotional validation and dismissal.

1. Narrative of scenario: The parent records a 7-minute bedtime meltdown on their phone. They later upload it to KidTalk Mirror. The LLM automatically transcribes and analyzes the interaction, distinguishing emotional attunement (“This feels big right now”) from dismissive reassurance (“You’re fine, stop crying”). Behind the scenes, the model runs emotion-recognition and developmental-language analysis, labeling moments that may activate shame triggers or build trust signals. It reconstructs the emotional timeline and generates concrete swaps, each explained with child-development rationale. The parent receives an Insight Report with tags and links to a “Shame vs. Guilt” Skill Builder article.

3. Step-by-step interaction
- Parent taps “New Report → Upload Audio File.”
- Chooses recorded bedtime clip; tags as “Bedtime.”
- Uploads, LLM transcribes and detects speaker roles.
- The model identifies emotion-laden phrases and classifies tone.
- Report generated:
  1. Gentle Summary (“You stayed patient during an emotional moment.”)
  2. Tags for Emotional Attunement / Shame Trigger.
  3. 2–3 swaps (e.g., “Instead of ‘Stop crying,’ try ‘I know this is hard, and I’m here.’”).
  4. Link to Skill Builder reading on validating emotions.
The system logs data into the dashboard for ongoing pattern tracking.

4. Data description

Needed data:
- Audio input: 5–10 min parent-child bedtime conversation.
- Pretraining corpus: Emotionally rich dialogues annotated for empathy, comfort, validation.
- Model resources: Fine-tuned cognitive-emotion detection models.
  
Data created:
- Annotated transcript with emotion tags.
- Generated swap suggestions + explanatory rationales.
- Engagement log for article view linked to “Shame Trigger” tag.

5. Evaluation
- Success: ≥70% of users rate feedback “gentle” and “useful.”
- Complications: Crying overlaps may cause transcription errors, may be handled by noise-tolerant speech segmentation.




### Scenario 3: Drive-Home Decompression Car Chat Analysis

Short description: Quick car-ride upload for hands-free insight into motivational tone.

1. Narrative of scenario
A working parent uploads a 6-minute car ride chat from their phone. The LLM filters out background noise and identifies parent utterances like “Why didn’t you listen?” or “What part was hardest?” In the background, it segments the conversation into emotional zones (directive, reflective, empathetic). The model generates a summary showing where the parent shifted tone and calculates a motivation-support ratio comparing positive vs. critical phrasing. The parent receives two personalized swaps designed for time-stressed moments with concise and encouraging rephrasings.

2. Step-by-step interaction
- Tap “Upload Audio File.”
- Select “Daily Check-in” context; upload voice memo.
- LLM transcribes with noise filtering.
- Each statement is tagged for tone and intention.
- Model constructs an impact timeline visualizing emotional peaks.
- App displays:
  1. Gentle Summary.
  2. Tone map and swap suggestions.
  3. Feedback button (“Was this helpful?”).
  4. Trend data updates the dashboard automatically.

3. Data description

Needed data:
- Audio: 5–10 min natural recording (mobile upload).
- Training data: Conversational corpora from everyday family interactions with overlapping speech.
- Context label: “Daily Check-in.”

Data created:
- Time-stamped transcript with tone shifts.
- Motivation-support ratio metric.
- User feedback ratings for continuous model refinement.

4. Evaluation
- Success: Insight Report generated within 2 minutes, tags coherent despite noise.
- Complications: Child silence or environmental interference; low-confidence sections omitted automatically.


### Scenario 4: Checking Growth for a Month

Short description: Parents exploring long-term progress via the Growth Dashboard.
1. Narrative of scenario
After four weeks of recordings, the parent opens the Growth Dashboard. The LLM aggregates prior reports and builds a 30-day data summary of “helpful” vs. “to-reduce” patterns. It visualizes progress through simple bar and line graphs e.g., a 30% decrease in controlling language and 25% increase in empathy phrases.
Behind the scenes, a trend analysis algorithm computes changes across all prior reports and generates a text summary in supportive language (“You’ve built more curiosity in your responses this month”). The dashboard turns linguistic analytics into human-readable insights, fostering sustained motivation.

2. Step-by-step interaction
- User taps “Growth” tab.
- LLM queries all past reports and tags.
- Computes weekly pattern frequencies and displays visuals.
- Generates “Weekly Win” card (e.g., “+20% more autonomy support!”).
- Parent taps into earlier reports to review specific examples.
- Data continuously updates as new reports arrive.

3. Data description

Needed data:
- Aggregated reports and historical pattern tags.
- Metadata (timestamp, context, tone classifications).
- Trend analysis algorithms trained on anonymized user progress data.

Data created:
- Aggregated metrics for each 7-day period.
- Visualization-ready data arrays.
- Personalized growth summaries.

4. Evaluation
- Success metrics: Users report “motivated” after viewing dashboard. 80% comprehension without guidance.
- Complications and potential errors: Weeks with sparse data. This may be improved with encouragement text (“Every recording helps you grow”).

### Scenario 5: Parents Learning a Skill Before a Tough Conversation

Short description:Pre-learning a communication strategy, then measuring real-world application.

1. Narrative of scenario
Before addressing their child about a poor test result, the parent visits the Skill Builder Hub and reads an article titled “Building Motivation without Pressure.” The LLM acts as a micro-coach, prompting reflection (“What does motivation sound like in your home?”). Later, the parent records the actual conversation. The system transcribes and analyzes whether the parent incorporated newly learned skills (e.g., using curiosity questions instead of directives). In the background, the LLM matches report tags to Skill Builder concepts, quantifying improvement (e.g., “2 more autonomy-supportive phrases than last week”).

2. Step-by-step interaction
- Tap “Learn” tab → “Building Motivation.”
- LLM presents interactive article with quick self-check prompts.
- Parent records the next conversation tagged “Homework.”
- LLM cross-references tags from the report with Skill Builder topic.
- Report summary highlights improved use of motivational language.
- Dashboard adds learning-linked improvement note.

3. Data description

Needed data:
- Skill Builder article metadata (concept tags, timestamps).
- User learning log (topics read).
- Audio of subsequent interaction for follow-up analysis.
- Pretraining data linking concept → linguistic example pairs from pedagogy and psychology research.

Data created:
- Linkage data between learning content and improved tag scores.
- Time-based progress logs for educational reinforcement.
- Fine-tuning data points on concept transfer.

4. Evaluation
- Success: Parents who read Skill Builder content show 20% more “helpful” tags.
- Complications: Time gaps between learning and recording reduce detectability. The app may suggest quick follow-up prompts to measure parents’ improvement on communication skills.

### Scenario 6: Comparing Homework vs. Playtime Communication

Short description:Capturing multiple daily contexts to reveal tone shifts and stress patterns.

1. Narrative of scenario
The parent records two distinct interactions: one during a tense homework session, another during relaxed weekend play. The LLM independently analyzes both and compares tone distributions. It finds that controlling language occurs 60% more during homework than playtime. Behind the scenes, the system clusters context-tagged reports and calculates tone differentials across situations. The LLM composes an insight summary: “You’re naturally supportive during play. Try transferring that tone to learning time.” The comparison strengthens self-awareness without judgment.

2. Step-by-step interaction
Record Homework and Playtime sessions separately.
Upload each as a New Report with context labels.
LLM tags both transcripts and generates context-based summaries.
Model compares linguistic metrics across contexts.
Dashboard visualizes contrasts (bar chart: supportive vs. controlling ratios).
User reads contextual insight and sets intention for next week.

3. Data description
Needed data:
1. Multiple recordings with context metadata.
2. Pretraining data labeled by situational type (e.g., routine, play, study).
3. Model components for cross-context comparison.

Data created:
1. Context-specific transcripts and pattern averages.
2. Differential analysis dataset for internal fine-tuning.
3. Personalized cross-context comparison report.
   
5. Evaluation
- Success: Parent correctly interprets context contrast and identifies growth target.
- Complications: Uneven sample sizes by different contexts. Dashboard may adjust normalization automatically.
