## Haowei's Final Project Idea Pitch

My idea is to create an LLM Planning/Efficiency App that supports people to formulate self-regulated interaction with AI that helps them to avoid misuses of AI tools and achieve the goals they have. This LLM Planning App will help people to set personal, academic, and workplace goals but facilitates self-regulated learning under the AI era. It will ask students the desired outcome they hope to achieve and monitor their progress. When users first start, they will initiate a goal they want to achieve and create a weekly or long term plan. An AI chatbot embedded in the system will then initiate daily quick check in with students in pop up format to monitor whether their knowledge and skills is following the track of achieving a goal and if there is sign of overreliance on AI. AI will provide personalized knowledge checks that fit with the context and content of users’ goals. AI will also vary the tone, check in frequency based on how the user does in forming a healthy relation with AI. When the user manages self-regulated learning with AI, the interventions may fade away.
