# App workflow development and testing (Milestone #5)

## 1. Application Workflow

### **Introduction to the Workflow**

This workflow (below) visualizes the journey of a parent's voice from a stressful moment in the living room to a helpful insight on their phone screen. It breaks down the "KidTalk Mirror" product into a series of steps (or "nodes") that transform raw audio into psychology-backed parenting advice.

Think of this system as a digital assembly line:
1.  **Input:** The parent captures a messy, real-life conversation.
2.  **Processing:** The system cleans up the sound, figures out who is talking, and acts as a "psychology detective" to spot patterns.
3.  **Output:** The parent receives a gentle, private report with specific things they can say differently next time, while the system quietly tracks their long-term growth.

![mermaid](workflow_kidtalkmirror_mermaid.png)


### **The User Journey & Node Breakdown**

Here is how the product works step-by-step, explaining the function of each node in the diagram.

#### **Phase 1: Capturing the Moment**
The interaction begins when a parent realizes a conversation is going off the rails (like a homework battle) and decides to get help.

* **Node 1: User Input Interface**
    * **What it is:** The "Record" or "Upload" screen on the app.
    * **Function:** This is where the user records the conversation or uploads an existing file. Crucially, the user also selects a **Context Label** (e.g., "Homework," "Bedtime," or "Playtime"), which tells the system what kind of situation to expect.

#### **Phase 2: Making Sense of the Audio**
Once the file is uploaded, the "backend" takes over. The user sees a "Processing..." screen while these invisible steps happen.

* **Node 2: Audio Processing Unit**
    * **What it is:** The noise filter.
    * **Function:** Real life is noisy. This node removes background sounds—like a TV blaring or a dishwasher running—so the computer can hear the voices clearly.
* **Node 3: ASR (Automatic Speech Recognition) & Diarization Engine**
    * **What it is:** The transcriber and sorter.
    * **Function:** It turns the audio into text (ASR) and, most importantly, identifies **who is speaking** (Diarization). It separates the "Parent" voice from the "Child" voice so the AI analyzes the parent's reaction, not the child's tantrum.

#### **Phase 3: The "Brain" (Analysis)**
Now that the system has a clean script of the conversation, it analyzes the meaning behind the words.

* **Node 4: Psychometric Analyzer**
    * **What it is:** The psychology expert.
    * **Function:** This is the core intelligence. It scans the parent's words for specific psychological patterns, tagging moments as "Autonomy Supportive" (helpful) or "Controlling" (unhelpful) based on developmental research. It assigns "impact scores" (e.g., pressure vs. empathy) to different parts of the conversation.
* **Node 6: Skill & Content Retriever**
    * **What it is:** The librarian.
    * **Function:** If Node 4 detects a specific issue (e.g., a "Shame Trigger"), this node searches the app's library for a short "Skill Builder" article or tip that explains that specific concept.

#### **Phase 4: Delivering Feedback**
The analysis is packaged into something the parent can actually use.

* **Node 5: Insight & Swap Generator**
    * **What it is:** The coach / writer.
    * **Function:** It takes the patterns and writes the **Insight Report**. It generates a "Gentle Summary" (non-judgmental feedback) and creates specific **Language Swaps**—showing the parent exactly what they said and offering a better phrase to use next time (e.g., "Instead of 'Stop crying,' try 'I know this is hard'").

#### **Phase 5: Tracking Growth (Long-Term)**
The interaction is over, but the data is saved to help the parent see progress over time.

* **Node 7: Longitudinal Database**
    * **What it is:** The secure digital filing cabinet.
    * **Function:** Instead of just saving the text, this node saves the *metrics* (e.g., "On Dec 12, the parent was 20% more empathetic than last week"). This allows the app to track trends safely and privately.
* **Node 8: Dashboard Aggregator**
    * **What it is:** The progress report creator.
    * **Function:** When the parent checks the app a week later, this node pulls data from the database to show "Weekly Wins" or charts comparing different contexts (e.g., "You are more patient during Playtime than Homework").

## 2. Implementation and Testing of Workflow

We chose to implement and test the three use cases of Homework, Bedtime Tantrum and Ride talk, all of which go through the workflow of ***Audio Processing—ASR Diarization — Psychometric Analyzer — Cognitive Science Database (RAG) —Insight & Swap Generator*** to the final deliverable of Insight report. 

To test the workflow, we built a model on the Open WebUI. We **uploaded Cognitive Science-based principles/pedagogies as database/RAG** to instruct the model, and used our metaprompt as well as scenario specific prompts in the system prompt. Due to the technical constraints, our implementation test does not include the Audio processing steps, and is going to start when we already have the input conversation transcript and the label (context of either Homework, Bed Time or Ride talk).

<p><strong>Patterns (Neutral Observations):</strong></p>
<ul>
  <li>There were repeated urgent cues (“Let’s do this quickly”, “We don’t have all day”), corrections (“No! Come on…”, “Stop getting mixed up”), and push-back on emotions (“Stop crying”, “You haven’t earned a break”).</li>
  <li>The responses leaned toward getting through the work and correcting the child, with less space for feelings or self-direction.</li>
</ul>

<p><strong>Targeted Suggestions:</strong></p>

<p><strong>A. Tone & Language Adjustments</strong><br>
Try these small swaps to help lower stress and support autonomy, without losing your original intention:</p>

<ol>
  <li>
    <strong>From urgency to calm collaboration:</strong><br>
    Swap “Let’s do number 5 quickly” → “Let’s see how we can figure out number 5 together.”<br>
    Instead of “You should know this by now,” try “What part feels tricky right now? Let’s tackle it step by step.”
  </li>
  <br>
  <li>
    <strong>Validating emotion and persistence:</strong><br>
    Try “It’s okay to get mixed up—lots of people do. Let’s slow down for a sec.”<br>
    Shift “Stop crying and just focus” → “It’s really hard to think when we feel upset. Want to take 30 seconds to breathe with me?”
  </li>
  <br>
  <li>
    <strong>Supporting thinking, not just answers:</strong><br>
    Replace “Just do it” with “Walk me through your thinking—what do you notice about the numbers?”<br>
    Instead of correcting quickly, try “What’s your first idea?”
  </li>
</ol>

<p><strong>B. Perspective & Approach Shifts</strong></p>
<ul>
  <li>Pause and ask: “Do I want speed right now, or confidence and learning?” Slowing down can support both.</li>
  <li>Try interpreting hesitation or requests for breaks as distress rather than defiance.</li>
  <li>Experiment with short “mini-breaks” or offering limited choices to support cooperation.</li>
</ul>
