## Shiyang's Final Project Idea Pitch

My idea is to create a web-based platform that uses multiple large language models to support qualitative coding. The platform will allow researchers to upload textual data—such as interview transcripts, student responses, or reflection logs—and automatically generate codes from several LLMs at once.
By displaying the outputs side by side, the system will help users compare how different models interpret the same data. Researchers will be able to review, edit, and finalize the codes within the interface, ensuring human judgment remains central to the process.
The goal of this tool is to make qualitative analysis more efficient, transparent, and collaborative, while also enabling systematic comparison of coding reliability and interpretive depth across different language models.
